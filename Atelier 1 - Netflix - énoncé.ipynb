<h1> Exercice </h1>

Le dataset "Titres Netflix" est une compilation exhaustive de films et de séries télévisées disponibles sur Netflix, couvrant divers aspects tels que le type de titre, le réalisateur, les acteurs, le pays de production, l'année de sortie, la classification, la durée, les genres (répertoriés dans) et une brève description. Ce dataset est essentiel pour analyser les tendances du contenu Netflix, comprendre la popularité des genres et examiner la distribution du contenu à travers différentes régions et périodes.

Les colonnes : 
- show_id : Un identifiant unique pour chaque titre.
- type : La catégorie du titre, qui peut être 'Film' ou 'Série télévisée'.
- title : Le nom du film ou de la série télévisée.
- director : Le(s) réalisateur(s) du film ou de la série télévisée. (Contient des valeurs nulles pour certaines entrées, en particulier les séries télévisées où cette information peut ne pas être applicable.)
- cast : La liste des acteurs principaux du titre. (Certaines entrées peuvent ne pas avoir cette information.)
- country : Le pays ou les pays où le film ou la série télévisée a été produit.
- date_added : La date à laquelle le titre a été ajouté à Netflix.
- release_year : L'année de sortie originale du film ou de la série télévisée.
- rating : La classification par âge du titre.
- duration : La durée du titre, en minutes pour les films et en saisons pour les séries télévisées.
- listed_in : Les genres auxquels appartient le titre.
- description : Un bref résumé du titre.

### importation de donnée

```python
import pandas as pd
data = pd.read_csv('./netflix_titles.csv', encoding='latin1')
```

```python
# copie du dataframe
df = data.copy()
```

```python
# verification des colonnes
df.info()
```

```python
# suppression des colonnes UNNAMED
df = df[['show_id', 'type', 'title', 'director', 'cast', 'country', 'date_added','release_year', 'rating', 'duration', 'listed_in', 'description']]
df.head()
```

```python
# identifiant
print(len(df))
print(df.isnull().mean()*100)
```

```python
df.set_index('show_id',inplace=True)
```

```python
# colonne date_added
df['date_added'] = pd.to_datetime(df['date_added'].str.strip() , format='%B %d, %Y')
df.head()
```

### duration

```python
# la colonne type
print(df['type'].unique())
print(df['duration'].unique())
```

```python
df.groupby(['type','duration'])['duration'].count()
```

```python
# creation des colonnes 
import numpy as np
df['duration (movies)'] = np.nan
df.loc[df['type'] == 'Movie', 'duration (movies)'] = df.loc[df['type'] == 'Movie', 'duration'].str.replace(' min','').astype(float)
df['duration (movies)']
```

```python
df['seasons (Tv show)'] = np.nan
df.loc[df['type'] == 'TV Show', 'seasons (Tv show)'] = df.loc[df['type'] == 'TV Show', 'duration'].str.replace(' Seasons','').str.replace(' Season','').astype(float)
df['seasons (Tv show)']
```

### supression de la colonne duration

```python
df.drop(columns=['duration'], inplace=True)
df.info()
```

## colonne annexe

```python
print(df["country"].unique())
```

### on prend chaque pays

```python
df['countries'] = df['country'].str.split(', ')
countries_explode = df.explode('countries')
countries_explode = countries_explode['countries']
countries_explode.to_csv('countries_explode.csv')
```

```python
countries_explode.value_counts()
```

### on gere les listed_in

```python
df['category_type'] = df['listed_in'].str.split(', ')
category_explode = df.explode('category_type')
category_explode = category_explode['category_type']
category_explode.to_csv('category_explode.csv')
```

```python
category_explode.unique()
```

### on gere la colonne cast

```python
df['cast_list'] = df['cast'].str.split(', ')
cast_list = df.explode('cast_list')
cast_list = cast_list['cast_list']
cast_list.to_csv('cast_list.csv')
```

```python
cast_list
```

```python
df['listed_in_list'] = df['listed_in'].str.split(', ')
categories_exploded = df.explode('listed_in_list')
categories_exploded = categories_exploded['listed_in_list']
categories_exploded.to_csv('listed_in_list.csv')
```

### date added

```python
import datetime as dt
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
df['day_added'] = df['date_added'].dt.dayofweek
```

```python
df.head(5)
```

```python
df.drop(columns=['cast','countries','listed_in_list','cast_list'],inplace=True)
```
